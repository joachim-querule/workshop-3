```python

 # Fonction Scatter_Graph

def Graphique(Data_abscice="Valeur x", Data_ordonne="Valeur y", abscice_value="Donné absice", ordonne_value="Donné ordonné", Title_Graph="titre du graphique"):
    
    # Valeur par défaut
    
    # abscice_value = "Valeur x"
    # ordonne_value = "Valeur y"
    # Data_abscice = "Donné absice"
    # Data_ordonne = "Donné ordonné"
    # Title_Graph = "titre du graphique"


    plt.figure(figsize=(8, 6)) # Taille de la fenêtre du graphique
    plt.scatter( Data_abscice, Data_ordonne) # Valeur utilisé par le graphique

    # plt.set_cmap(alpha=0.8)
    
    plt.title = Title_Graph # Nom du Graph/Fenêtre_
    plt.xlabel =  abscice_value # Nom de l'axe X
    plt.ylabel =  ordonne_value # Nom de l'axe Y

    return plt.show()

```

```python
def Choice_feature():
    i = 1
    for value in features:
        print(f'{i}){value}')
        i+=1
    n = int(input(" entre le numéro de la feature"))
    print(f'Vous avez choisie {features[n-1]}')
    return features[n-1]
```
